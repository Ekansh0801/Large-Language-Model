#### Implementing a Tokenizer class

```python
class simpleTokenizerV1:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {value : key for key, value in vocab.items()}

  def encode(self, text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [token.strip() for token in preprocessed if token.strip()]

    ids = [self.str_to_int[token] for token in preprocessed]
    return ids

  def decode(self, ids):
    sentence = " ".join([self.int_to_str[id] for id in ids])
    sentence = re.sub(r'\s+([,.?!"()\'])', r'\1', sentence)
    return sentence
```

### Updating tokenizer class

```python
class simpleTokenizerV2:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {value : key for key, value in vocab.items()}

  def encode(self, text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [token.strip() for token in preprocessed if token.strip()]

    preprocessed = [item if item in self.str_to_int else '<|unk|>' for item in preprocessed]

    ids = [self.str_to_int[token] for token in preprocessed]
    return ids

  def decode(self, ids):
    sentence = " ".join([self.int_to_str[id] for id in ids])
    sentence = re.sub(r'\s+([,.?!"()\'])', r'\1', sentence)
    return sentence
```

#### one of the most easiest and intutive way to create input - output pairs for the next-word prediction task is the create two variables x for input and y for output which is just the input shifted by 1

#### context size represents size of input as well as how many predictions will we make in one iteration

```python
context_size = 4 # length of input

# The context size of 4 means the model is trained to look at sequence of 4 words (or tokens)
# to predict the next token in the sequence

#The input x is the first 4 tokens [1, 2, 3, 4], and the target y is the next 4 tokens [2, 3, 4, 5]

x = enc_sample[:context_size]
y = enc_sample[1:context_size + 1]
```

```python
print(x)
print(y)
```

Processing the inputs along with the targets, which are the inputs shifted by one position,
we can then create the next-word prediction tasks as

```python
for i in range(1, context_size + 1):
  context = enc_sample[:i]
  desired = enc_sample[i]

  print(context, "--->", desired)
```

Everything left of the arrow (---->) refers to the input an LLM would receive, and the token
ID on the right side of the arrow represents the target token ID that the LLM is supposed to
predict.

```python
for i in range(1, context_size + 1):
  context = enc_sample[:i]
  desired = enc_sample[i]

  print(tokenizer.decode(context), "--->", tokenizer.decode([desired]))
```

## Implementing a Data Loader

Step 1: Tokenize the entire text
    
Step 2: Use a sliding window to chunk the book into overlapping sequences of max_length

Step 3: Return the total number of rows in the dataset

Step 4: Return a single row from the dataset

```python
from torch.utils.data import Dataset, DataLoader

class GPTDatasetV1(Dataset):
  def __init__(self, txt, tokenizer, max_length, stride):
    self.input_ids = []
    self.output_ids = []

    # Tokenize the entire text
    token_ids = tokenizer.encode(txt, allowed_special={'<|endoftext|>'})

    # use a sliding window to chunk the book into overlapping sequence of maximum length
    for i in range(0, len(txt) - max_length, stride):
      input_chunk = token_ids[i:i + max_length]
      output_chunk = token_ids[i + 1: i + max_length + 1]

      self.input_ids.append(torch.tensor(input_chunk))
      self.output_ids.append(torch.tensor(output_chunk))


  def __len__(self):
    return len(self.input_ids)


  def __getitem__(self, idx):
    return self.input_ids[idx], self.output_ids[idx]


```


### The GPTDatasetV1 class in listing 2.5 is based on the PyTorch Dataset class.

It defines how individual rows are fetched from the dataset.

Each row consists of a number of
token IDs (based on a max_length) assigned to an input_chunk tensor.

The target_chunk
tensor contains the corresponding targets.

The following code will use the GPTDatasetV1 to load the inputs in batches via a PyTorch

Step 1: Initialize the tokenizer

Step 2: Create dataset

Step 3: drop_last=True drops the last batch if it is shorter than the specified batch_size to prevent loss spikes
during training

Step 4: The number of CPU processes to use for preprocessing

```python
def create_dataloader_v1(txt, batch_size = 4, max_length = 256, stride = 128, shuffle=True, drop_last = True, num_workers = 0):


  # initialize the tokenizer
  tokenizer = tiktoken.get_encoding("gpt2")

  # creeate the dataset
  dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

  # create a dataloader
  dataloader = DataLoader(
      dataset,
      batch_size=batch_size,
      shuffle = shuffle,
      drop_last = drop_last,
      num_workers=num_workers
  )

  return dataloader
```

Convert dataloader into a Python iterator to fetch the next entry via Python's built-in next() function
