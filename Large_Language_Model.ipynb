# Building Large Language Model from Scratch

## Stage - 1 : Data Preprocessing

```python
# Loading the Dataset

with open('the-verdict.txt', 'r', encoding='utf-8') as file:
  raw_text = file.read()

print(len(raw_text))
```

```python
# Printing first 100 characters

print(raw_text[:100])
```

### Implementing Tokenization

#### step - 1: Creating Tokens

```python
# Importing re library for tokenization purpose
import re

# splitting the text on any special character or -- or space
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)

# storing only non whitespace characters
preprocessed = [item.strip() for item in preprocessed if item.strip()]

print(len(preprocessed))

print(preprocessed[:100])
```

#### Step - 2: assigning token ID's

```python
all_words = sorted(set(preprocessed))

vocab_size = len(all_words)
print(vocab_size)
```

```python
vocab = {token:integer for integer, token in enumerate(all_words)}
# print(vocab)
```

```python
for i, item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break
```

#### Implementing a Tokenizer class

```python
class simpleTokenizerV1:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {value : key for key, value in vocab.items()}

  def encode(self, text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [token.strip() for token in preprocessed if token.strip()]

    ids = [self.str_to_int[token] for token in preprocessed]
    return ids

  def decode(self, ids):
    sentence = " ".join([self.int_to_str[id] for id in ids])
    sentence = re.sub(r'\s+([,.?!"()\'])', r'\1', sentence)
    return sentence
```

```python
# check if the class is working succesfully

texty = """Well!--even through the prism of Hermia's tears I felt able to face the fact with equanimity.
Poor Jack Gisburn! The women had made him--it was fitting that they should mourn him.
Among his own sex fewer regrets were heard, and in his own trade hardly a murmur.
Professional jealousy? Perhaps. If it were, the honour of the craft was vindicated by little Claude Nutley,
who, in all good faith, brought out in the Burlington a very handsome "obituary" on Jack--one of those
showy articles stocked with random technicalities that I have heard (I won't say by whom) compared to
Gisburn's painting. And so--his resolve being apparently irrevocable--the discussion gradually
died out, and, as Mrs. Thwing had predicted, the price of "Gisburns" went up."""

tokenizer = simpleTokenizerV1(vocab)

ids = tokenizer.encode(texty)
print(ids)
```

```python
decoded_text = tokenizer.decode(ids)
print(decoded_text[:100])
```

### Updating tokenizer class

```python
class simpleTokenizerV2:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {value : key for key, value in vocab.items()}

  def encode(self, text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [token.strip() for token in preprocessed if token.strip()]

    preprocessed = [item if item in self.str_to_int else '<|unk|>' for item in preprocessed]

    ids = [self.str_to_int[token] for token in preprocessed]
    return ids

  def decode(self, ids):
    sentence = " ".join([self.int_to_str[id] for id in ids])
    sentence = re.sub(r'\s+([,.?!"()\'])', r'\1', sentence)
    return sentence
```

#### checking the new tokenizer class

```python
dec_text = tokenizer2.decode(n_ids)
print(dec_text)
```

```python
import importlib
import tiktoken

print("Tiktoken version:", importlib.metadata.version("tiktoken"))
```

#### using the GPT-2 tokenizer

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

<div class="alert alert-block alert-success">
The usage of this tokenizer is similar to SimpleTokenizerV2 we implemented previously via
an encode method:</div>



```python
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
     "of someunknownPlace."
)

integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(integers)
```

<div class="alert alert-block alert-success">
We can then convert the token IDs back into text using the decode method, similar to our
SimpleTokenizerV2 earlier:</div>


```python
strings = tokenizer.decode(integers)

print(strings)
```

**Let us take another simple example to illustrate how the BPE tokenizer deals with unknown tokens**

```python
integers = tokenizer.encode("Swati will be mine!!!")

print(integers)

strings = tokenizer.decode(integers)
print(strings)
```
