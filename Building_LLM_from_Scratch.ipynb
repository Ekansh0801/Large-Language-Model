#### Implementing a Tokenizer class

```python
class simpleTokenizerV1:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {value : key for key, value in vocab.items()}

  def encode(self, text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [token.strip() for token in preprocessed if token.strip()]

    ids = [self.str_to_int[token] for token in preprocessed]
    return ids

  def decode(self, ids):
    sentence = " ".join([self.int_to_str[id] for id in ids])
    sentence = re.sub(r'\s+([,.?!"()\'])', r'\1', sentence)
    return sentence
```

```python
class simpleTokenizerV2:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {value : key for key, value in vocab.items()}

  def encode(self, text):
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [token.strip() for token in preprocessed if token.strip()]

    preprocessed = [item if item in self.str_to_int else '<|unk|>' for item in preprocessed]

    ids = [self.str_to_int[token] for token in preprocessed]
    return ids

  def decode(self, ids):
    sentence = " ".join([self.int_to_str[id] for id in ids])
    sentence = re.sub(r'\s+([,.?!"()\'])', r'\1', sentence)
    return sentence
```

## Implementing a Data Loader

Step 1: Tokenize the entire text
    
Step 2: Use a sliding window to chunk the book into overlapping sequences of max_length

Step 3: Return the total number of rows in the dataset

Step 4: Return a single row from the dataset

```python
from torch.utils.data import Dataset, DataLoader

class GPTDatasetV1(Dataset):
  def __init__(self, txt, tokenizer, max_length, stride):
    self.input_ids = []
    self.output_ids = []

    # Tokenize the entire text
    token_ids = tokenizer.encode(txt, allowed_special={'<|endoftext|>'})

    # use a sliding window to chunk the book into overlapping sequence of maximum length
    for i in range(0, len(txt) - max_length, stride):
      input_chunk = token_ids[i:i + max_length]
      output_chunk = token_ids[i + 1: i + max_length + 1]

      self.input_ids.append(torch.tensor(input_chunk))
      self.output_ids.append(torch.tensor(output_chunk))


  def __len__(self):
    return len(self.input_ids)


  def __getitem__(self, idx):
    return self.input_ids[idx], self.output_ids[idx]


```


### The GPTDatasetV1 class in listing 2.5 is based on the PyTorch Dataset class.

It defines how individual rows are fetched from the dataset.

Each row consists of a number of
token IDs (based on a max_length) assigned to an input_chunk tensor.

The target_chunk
tensor contains the corresponding targets.

The following code will use the GPTDatasetV1 to load the inputs in batches via a PyTorch

Step 1: Initialize the tokenizer

Step 2: Create dataset

Step 3: drop_last=True drops the last batch if it is shorter than the specified batch_size to prevent loss spikes
during training

Step 4: The number of CPU processes to use for preprocessing

```python
def create_dataloader_v1(txt, batch_size = 4, max_length = 256, stride = 128, shuffle=True, drop_last = True, num_workers = 0):


  # initialize the tokenizer
  tokenizer = tiktoken.get_encoding("gpt2")

  # creeate the dataset
  dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

  # create a dataloader
  dataloader = DataLoader(
      dataset,
      batch_size=batch_size,
      shuffle = shuffle,
      drop_last = drop_last,
      num_workers=num_workers
  )

  return dataloader
```

# Stage - 2 : ATTENTION MECHANISM

### We will now implement a simplified attention mechanism(without trainable weights) for strengthning our understanding

Simplified Attention Mechanism

lets do for a simple example first then will do for the whole input we have generated in previous part

```python
import torch

inp = torch.tensor(
    [
        [0.43, 0.15, 0.89],  # your
        [0.55, 0.87, 0.66],  # journey
        [0.57, 0.85, 0.64],  # starts
        [0.22, 0.58, 0.33],  # with
        [0.77, 0.25, 0.10],  # one
        [0.05, 0.80, 0.55]   # step
    ]
)
```

##### we will find out context word for all words relative to every other word but for now let's do for journy to understand the process

#### and we will use the property that dot product finds relativity between different words

#### since we are focusing on only word for now we will call it query

```python
query = inp[1]

# initializing an empty tensor for storing attention weights
attn_scores2 = torch.empty(inp.shape[0])


# looping in the input tensor to find dot product with every word
for i, xi in enumerate(inp):
  attn_scores2[i] = torch.dot(query, inp[i])

print(attn_scores2)
```

### but these scores are hard to analyze for a person so let's normalize them so that they can be intrepretated as probability and we can directly compare which word is most similar

#### We can normalize by dividing each value by total sum

```python
attn_wts_norm1 = attn_scores2 / torch.sum(attn_scores2)

# straightforward normalizing by dividing each value by total sum
print(attn_wts_norm1)
print(attn_wts_norm1.sum())
```

### But there will be many limitations with this one which we have seen in logistic regression for multiple classes hence like that here also we switch to softmax normalization

```python
def softmax_naive(x):
  return torch.exp(x) / torch.exp(x).sum(dim = 0)

attn_wts_norm2 = softmax_naive(attn_scores2)

print(attn_wts_norm2)
print(attn_wts_norm2.sum())
```

#### since sum is one we can now easily say  that the query(journey) is closest to journey and starts(word 2 and word 3) since they have highest attention scores respect to others

#### now to get the context vector for word journy we will multiply the calculate attention weights with the vector embeddings and sum all to get a final context vector

```python
query = inp[1]

context_vec_2 = torch.zeros(query.shape)

for i, xi in enumerate(inp):
  context_vec_2 += xi * attn_wts_norm2[i]


print(context_vec_2)
```

### this is the context vector for query with respect to all other tokens this we calculated just for understanding now lets calculate for all the tokens with respect to all tokens

```python
n = inp.shape[0]
attn_scores = torch.empty(n, n)


for i,x_i in enumerate(inp):
  for j, x_j in enumerate(inp):
    attn_scores[i,j] = torch.dot(x_i,x_j)

print(attn_scores)
```

#### for calculating attention scores we used for loops but they are very slow but we can achieve the same result by multiplying inp x inp.T we can verify that by checking and result will also be same

```python
attn_scores = inp @ inp.T
print(attn_scores)
```

#### this is faster than for loops

### now we need to normalize each row

```python
attn_weights = torch.softmax(attn_scores, dim = 1) # normalizing across each row hence dim = 1 as sum all columns for a row

print(attn_weights)
```

#### lets verify normalization

```python
print(attn_weights.sum(dim = 1))
```

### lets now find the context vector matrix that contains context vectors for all the tokens/words

```python

context_matrix = attn_weights @ inp

print(context_matrix)
```

#### we could have done with for loops but that will be slower and this is faster and we can verify it that it will be exactly same even the dimension will be same as final context matrix must be 6 x 3 so that comes after multipling weights_matrx * input matrix = (6 x 6) X (6 x 3) = (6 x 3) which is the size of context_matrix also we can see for eg row1 of att_weights( att_weights of token 1 wrt to all token ) and whole inp matrx by row * row_wise multiplication of matrix each row is context vector when attn_weights of each word is multiplied by its vector

## self attention mechanism with trainable weights

### in this section we will implement self attention mechanism used in original transformer architecture, the GPT models and some famous LLM's. This self attention mechanism is also known as "Scaled dot-product attention".

```python
import torch

inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your     (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts   (x^3)
   [0.22, 0.58, 0.33], # with     (x^4)
   [0.77, 0.25, 0.10], # one      (x^5)
   [0.05, 0.80, 0.55]] # step     (x^6)
)
```

```python
x_2 = inputs[1]
d_in = inputs.shape[1]
d_out = 2
```

#### we can see from the output we have successfully projected input form 3-d space to a 2-d space

#### now we will forget the input vectors and will only use the query, key and value matrices

#### will do something similar like we did in simplified attention weights without weights there we use input itself and multiplied by input to find attention scores and then attention weights and then multiplied by input itself to calculate the final context vector

#### here we will use query as a query matrix to search and key matrix to find the attention scores of all inputs and then will multiply the attention weidhts with value to get final context vector

#### so like previously lets check for a single query then will do for whole matrix

```python
query_2 = query[1]

attn_score_2 = query_2 @ keys.T

print(attn_score_2)
```

#### this is basically attention scores of token2 (journey ) with all tokens

#### because query_2 is 1 x 2 and keys is 6 x 2 so we need to take transpose of keys to get the attention score vector of query_2

#### similarly lets find the attention score matrix for all querys

```python
attention_scores = query @ keys.T
print(attention_scores)
```

#### lets now convert these attention scores into attention weights

#### but this time we will not apply softmax directly this time we will scale the values by root(dimension of keys) then will apply softmax.

```python
dim_keys = keys.shape[-1]

attn_weights_2 = torch.softmax(attention_scores / dim_keys ** (0.5), dim = -1)

print(attn_weights_2)
```

```python
print(attn_weights_2.sum(dim = 1))
```

### But why do we here divide by SQRT(dimension)?

Reason 1 : For stability in learning

The softmax function is sensitive to the magnitudes of its inputs. When the inputs are large, the difference between the exponential values of each input becomes much more pronounced. This causes the softmax output to become "Peaky", where the highest value recieves almost all the probability mass, and the rest recieves very little.

In attention mechanisms, particularly in transformers, if the dot product between query and key becomes too large (like multiplying by 8 in the example), the attention score becomes very large. This results in a very sharp softmax distribution, making the model confident in one particulay "key". Such sharp distributions can make learning unstable.

```python
import torch

tensor = torch.tensor([0.1, -0.2, 0.3, -0.2, 0.5])


# apply softmax without scaling
softmax_result = torch.softmax(tensor, dim = -1) # -1 for last dimension as column is last dimension
print("Softmax before scaling:", softmax_result)

# multiply by 8 and then applying softmax
scaled_tensor = 8 * tensor
softmax_scaled_result = torch.softmax(scaled_tensor, dim = -1)
print("Softmax After scaling:", softmax_scaled_result)
```

#### we can see most of the weight is assingened to last component after scaling which is not good thats why we divide by SQRT(dim_keys)

### But why SQRT(dim_keys) ?

Reason - 2: To make the variance of dot product stable

The dot product of Q and K increases the variance because multiplying two random numbers.

The increase in dimension increase the variance

Diving by sqrt(dim) keeps variance close to 1

So thats the reason we divide it by SQRT(dim) before applying softmax and thats why it is called *** Scaled dot product attention ***

```python
dim_keys = keys.shape[1]
attention_weights = torch.softmax(attention_scores / dim_keys ** 0.5, dim = 1)

print(attention_weights)
```

```python
print(attention_weights.sum(dim = 1))
```

#### now lets calculate the context vectors of each input token by multiplying attention_weights x values matrix since attention weights is 6x6 and values is 6x2 we should get 6x2 as final context_matrix

```python
context_matrix = attention_weights @ value

print(context_matrix)
```

### Let's implement a class for doing all this as we will not write code seperately for each section we will create a class

```python
import torch.nn as nn

class Self_Attention_v1(nn.Module):

  def __init__(self, in_dim, out_dim):
    super().__init__()
    self.W_query = nn.Parameter(torch.rand(in_dim, out_dim))
    self.W_key = nn.Parameter(torch.rand(in_dim, out_dim))
    self.W_value = nn.Parameter(torch.rand(in_dim, out_dim))


  def forward(self, x):
    keys = x @ self.W_key
    queries = x @ self.W_query
    values = x @ self.W_value

    attn_scores = queries @ keys.T
    attn_weights = torch.softmax(attn_scores / keys.shape[1] ** 0.5, dim = -1)

    context_vec = attn_weights @ values
    return context_vec
```

```python
torch.manual_seed(123)

sa_v1 = Self_Attention_v1(d_in, d_out)
print(sa_v1.forward(inputs))
```

#### We can improve Self_Attention_V1 implementation further by utilizing Pytorch nn.Linear Layers

```python
import torch.nn as nn

class Self_Attention_v2(nn.Module):

  def __init__(self, in_dim, out_dim, qkv_bias = False):
    super().__init__()
    self.W_query = nn.Linear(in_dim, out_dim, bias = qkv_bias )
    self.W_key = nn.Linear(in_dim, out_dim, bias = qkv_bias )
    self.W_value = nn.Linear(in_dim, out_dim, bias =qkv_bias )


  def forward(self, x):
    keys = self.W_key(x)
    queries = self.W_query(x)
    values = self.W_value(x)

    attn_scores = queries @ keys.T
    attn_weights = torch.softmax(attn_scores / keys.shape[-1] ** 0.5, dim = -1)

    context_vec = attn_weights @ values
    return context_vec
```

```python
torch.manual_seed(789)

sa_v2 = Self_Attention_v2(d_in, d_out)
print(sa_v2.forward(inputs))
```

### Now we will look at CAUSAL ATTENTION

* It is also known as Masked Attention
* It restricts model to only consider previous and current inputs in a sequence, when processing any given token
* This is in contrast to self attention mechanism, which allows access to the entire input sequence at once.
* When computing attention scores, the causal attention mechanism ensures that the model only factors in tokens that occur at or before the current token in sequence
* To achieve this in GPT like models, we mask out the future tokens, which come after the current token in the given input sequence.

#### we mask out the entries above the diagonal in attention weight matrix and normalize the remaining values to get masked attention weights

Applying a Causal Attention Mask

Strategy -> Get Attention Weights then zero out the entries above the main diagonal and normalize the result matrix(each row)

```python
inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your     (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts   (x^3)
   [0.22, 0.58, 0.33], # with     (x^4)
   [0.77, 0.25, 0.10], # one      (x^5)
   [0.05, 0.80, 0.55]] # step     (x^6)
)
```

```python
queries = sa_v2.W_query(inputs)
keys = sa_v2.W_key(inputs)

attn_scores = queries @ keys.T
attn_weights = torch.softmax(attn_scores / keys.shape[-1] ** 0.5, dim = -1)
print(attn_weights)
```

### We can now use PyTorch's Tril function to create a mask where the values above the diagonals are zero

```python
context_length = attn_scores.shape[0]

mask_simple = torch.tril(torch.ones(context_length, context_length))
print(mask_simple)
```

#### we will now multiply with attention weight matrix so that all upper traingular values are zero

```python
masked_simple = attn_weights * mask_simple
print(masked_simple)
```

```python
row_sums = masked_simple.sum(dim = 1, keepdim=True)
# print(row_sums)

masked_simple_norm = masked_simple / row_sums

print(masked_simple_norm)
```

#### We can now use this and multiply with values matrix to get context vectors and succesfully say we have implemented causal attention

### But if we see that when applying softmax during attention score normalization to get attention weights we are essentially dividing each entry by sum of current row so anyway the future entries have somewhat utilized so we can't really say we have masked future tokens.

#### and this is known as data leakage problem

### There is a more efficient way

Calculate Attention Scores  --> Apply upper triangular infinity mask  --> Apply softmax

When we take upper triangular infinity mask then entries above the diagonal are -inf then if we apply softmax then anywasy exp(-inf) is 0 so the next tokens did not affect the previous tokens in any manner.

```python
print(attn_scores)
```

```python
mask = torch.triu(torch.ones(context_length, context_length), diagonal = 1)

print(mask)

masked = attn_scores.masked_fill(mask.bool(), -torch.inf)
# masked_fill those entries will -inf where there is 1 thats why we took an upper traingular matrix this time
print(masked)
```

#### now  we will apply softmax to get attention weights

```python
attn_weights = torch.softmax(masked / keys.shape[-1] ** 0.5, dim = 1)
print(attn_weights)
```

### The following causalattentionclass is similar to selfattention class except that we have added dropout and causal mask

Step 1: Compared to the previous SelfAttention_v1 class, we added a dropout layer.
    
Step 2: The register_buffer call is also a new addition (more information is provided in the following text).

Step 3:  We transpose dimensions 1 and 2, keeping the batch dimension at the first position (0).

Step 4: In PyTorch, operations with a trailing underscore are performed in-place, avoiding unnecessary memory
copies

```python
class CausalAttention(nn.Module):

  def __init__(self, d_in, d_out, context_length, dropout, qkv_bias=False):
    super().__init__()
    self.d_out = d_out
    self.d_in = d_in
    self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
    self.W_key   = nn.Linear(d_in, d_out, bias=qkv_bias)
    self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)
    self.dropout = nn.Dropout(dropout)   # new
    self.register_buffer('mask', torch.triu(torch.ones(context_length, context_length), diagonal = 1))   # new


  def forward(self, x):
    b, num_tokens, d_in = x.shape # New batch dimension b = no of batches
    keys = self.W_key(x)
    queries = self.W_query(x)
    values = self.W_value(x)


    attn_scores = queries @ keys.transpose(1, 2)  # transpose only inside a batch because (2,6,3 must become 2, 3, 6)
    attn_scores.masked_fill_(self.mask.bool()[:num_tokens, :num_tokens], -torch.inf)
    attn_weights = torch.softmax(attn_scores / keys.shape[-1] ** 0.5, dim = -1)
    attn_weights = self.dropout(attn_weights)

    context_vec = attn_weights @ values
    return context_vec
```

The use of register_buffer in
PyTorch is not strictly necessary for all use cases but offers several advantages here.

For
instance, when we use the CausalAttention class in our LLM, buffers are automatically
moved to the appropriate device (CPU or GPU) along with our model, which will be relevant
when training the LLM in future chapters.

This means we don't need to manually ensure
these tensors are on the same device as your model parameters, avoiding device mismatch
errors.

```python
torch.manual_seed(123)
context_length = batch.shape[1]

ca = CausalAttention(d_in, d_out,context_length, 0.0)
context_vecs = ca(batch)

print(context_vecs)
print(context_vecs.shape)
```

### Multi Head Attention

#### The term multi head attention refers to dividing attention mechanism into multiple heads each operating independently.

a) stacking multiple single head attention layers
  1) Implementing multi-head attention involves creating multiple instances of the self-attention mechanism; each with it's own weights, and combining their outputs.
  2) Can be computationally intensive, but make LLM's powerful at complex pattern recognition tasks.

Previously in single head we took example of input 6x3 and converted into context vector of 6x2 here suppose we have currently 2 multi heads each having its own Wq, Wk, Wv matrix and then each will have its own keys, queries and values matrix and each will have context matrix of 6x2 and in multi - head attention we just concatenate the result along the columns of all the heads so here the final context vector will be 6x4

In practical terms, implementing multi- head attention involves creating multiple instances of self - attention mechanism, each with its own weights and combining their outputs.

In code we can achieve this by implementing a simple MultiHeadWrapperClass that stacks multiple instances of of our previously implemented CausalAttentionModule

```python
class MultiHeadAttentionWrapper(nn.Module):
  def __init__(self, d_in, d_out, context_length, dropout, num_heads, qkv_bias = False):
    super().__init__()
    self.heads = nn.ModuleList([CausalAttention(d_in, d_out, context_length, dropout, qkv_bias) for _ in range(num_heads)])

  def forward(self, x):
    return torch.cat([head(x) for head in self.heads], dim = -1)
```

We implemented a MultiHeadAttentionWrapper that combined multiple single head attention modules.
However, note that these are processed sequentially via [head(x) for head in self.heads] in the forward method

We can improve this implementation by processing heads in parallel
one way to achieve that is by computing the outputs of all attention heads simultaneously  via matrix multiplication.

For information GPT - 3 has 96 attention heads

#### The problem with this approach is that since GPT-3 has 96 attention heads so we will have 96 different weight matrices for Wkey, Wqueries, Wvalues and to get another 96 keys queries and values matrices we need 96 matrix multiplications which will be very time comsuming

#### So we need some method to reduce number of matrix multiplication

### Implementing Multi Head Attention with weight splits

#### Instead of maintaing seperate classes for causal attention and multiheadattentionwrapper, We combine both into one single MultiHeadAttentionClass

Currently we saw each attention head had a dimension - 2

number of attention - heads is specified
so Head dimension = d_out / # heads

so d_out = head dimension * (# heads)

here for eg head dimension = 2 and number of heads we took = 2
so head_dimension = 2 * 2 = 4

Instead of maintaining two seperate classes, we can combine both of them into one single class

In the multiheadattentionwrapper, multiple attention heads are implemented by creating a list of CausalAttention objects (self.heads), each representing a seperate attention head.

The causal attention class independently performs the attention mechanism, and results from each head are concatenated.

In contrast, the following MultiHeadAttention Class integrates the multi-head functionality with a single class

It splits the input into multiple heads by reshaping the projected query, keys and values tensors and then combines the results from these heads after computing attention.

suppose we have for eq shape of input = (1, 3, 6) means number of batch = 1, tokens = 3, input_dim = 6  and we suppose output_dim = 6(which is generally in gpt models d_in = d_out) so Wq, Wk, Wv will be 6x6 matrix and hence keys, queries and values will be 1 X 3 X 6

of course the number of dimensions will be very huge in gpt we are taking small for illustration

once we get the keys, quries and values matrices suppose 1x3x6 here 6 is the output dim and head_dim = output_dim / # heads

so we will now unroll the output_dim = # heads x head_dim

here suppose number of heads = 2 so 1x3x2x3

```python
class MultiHeadAttention(nn.Module):
  def __init__(self, d_in, d_out, context_length, dropout, num_heads, qkv_bias = False):
    super().__init__()
    assert (d_out % num_heads == 0), \
      "d_out must be divisible by num_heads"

    self.d_out = d_out
    self.num_heads = num_heads
    self.head_dim = d_out // num_heads


    self.W_query = nn.Linear(d_in, d_out, bias = qkv_bias)
    self.W_key = nn.Linear(d_in, d_out, bias = qkv_bias)
    self.W_value = nn.Linear(d_in, d_out, bias = qkv_bias)


    self.out_proj = nn.Linear(d_out, d_out)  # linear layer to combine head outputs
    self.dropout = nn.Dropout(dropout)
    self.register_buffer("mask", torch.triu(torch.ones(context_length, context_length), diagonal = 1))


  def forward(self, x):
    b, num_tokens, d_in = x.shape

    keys = self.W_key(x)
    queries = self.W_query(x)
    values = self.W_value(x)

    # we implicityly split them matrix by adding a 'num_heads' dim
    # unroll last dim: (b, num_tokens, out_dim) -> (b, num_tokens, num_heads, head_dim)

    keys = keys.view(b, num_tokens, self.num_heads, self.head_dim)
    queries = keys.view(b, num_tokens, self.num_heads, self.head_dim)
    values = keys.view(b, num_tokens, self.num_heads, self.head_dim)

    # now we have dim (b, num_tokens, num_heads, head_dim)
    # but if we change to (b, num_heads, num_tokens, head_dim) then we will have the matrix as (1,2,3,3) means for each batch we have now seperate sum_tensor of each
    # head so we can easily calculate the attention scores of each head independently and simultaneouly

    keys = keys.transpose(1,2)
    queries = queries.transpose(1,2)
    values = values.transpose(1,2)

    # now we will find out the attention scores usign queries and keys matrix and remember we find it using queries * keysTranspose
    # so here also we will multiply queries * keysT but the last two dimensions will be transposed so that we get attention scores for each each batch and each head
    # so new_dim will be (num_batch, num_heads, num_tokens, num_tokens)
    attn_score = queries @ keys.transpose(2,3)


    # now we will move towards finding attention_weights
    # firstly we will implement causal attention


    # masking the tokens after the current token by replacing with -inf
    mask_bool = self.mask.bool()[:num_tokens, :num_tokens]

    # use the mask to fill attention_scores
    attn_score.masked_fill_(mask_bool, -torch.inf)

    # now apply softmax and before that divide by sqrt(head_dim) why? we saw in causal attention

    attn_weights = torch.softmax(attn_score / keys.shape[-1] ** 0.5, dim = -1)

    #  can apply dropout also
    attn_weights = self.dropout(attn_weights)


    # now to get context vector we will multiply attn_weights * values
    #                                            (num_batch, num_heads, num_tokens, num_tokens) * (num_batch, num_heads, num_tokens, head_dim)
    #                                           => (num_batch, num_heads num_tokens, head_dim)



    context_vectors = attn_weights @ values

    # but our goal was to get context_vector in form of (num_batch, num_tokens, d_out)
    # so will we swap dimensions and merge

    # swap dimensions
    context_vectors = context_vectors.transpose(1,2)

    # new_dim = (num_batch, num_tokens, num_heads, head_dim)
    # and now we will merge back into (num_batch, num_tokens, d_out = (num_heads * head_dim))

    context_vectors = context_vectors.contiguous().view(b, num_tokens, self.d_out)
    # we need the context vectors to be stored in contigous memory location

    context_vectors = self.out_proj(context_vectors)   # optional projection


    return context_vectors


```
